# Univariate Analysis

# 1. Pie chart

```python
days = tips['day'].value_counts().sort_index()
print(days)
days_label = ['Fri','Sat', 'Sun','Thur']
plt.pie(days, labels = days_label, autopct = '%0.2f%%')
plt.show()
```

# 2. Count plot

```python
sns.countplot(data = titanic , x = 'Survived')
```

# 3. Kde + Rug plot

```python
sns.kdeplot(data = titanic, x = 'Fare')
sns.rugplot(data = titanic, x = 'Fare')
plt.show()
```

# 4. Violin plot

```python
sns.violinplot(titanic['Embarked'])
```

# 5. Box plot

```python
sns.boxplot(data = titanic, x = 'Age')
```

# 6. Boxen plot

```python
sns.boxenplot(data = titanic, x = 'Age')
```

# 7. Histogram

```python
total_bill = tips['total_bill']

sns.histplot(total_bill,
         bins = 15,
         color = 'black',
         edgecolor = 'white'
        )
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = iris, x = 'sepal_length', y = 'petal_width', hue = 'species' )
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(titanic, hue = 'Survived')
```

# 3. Heatmap

```python
sns.heatmap(iris.select_dtypes(include='number').corr(),cmap='coolwarm', annot = True)
plt.title('Heatmap on tips dataset')
plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(titanic,col = 'Sex',row = 'Embarked')
g.map(sns.scatterplot,'Survived','Pclass')
plt.show()
```

# 5. PairGrid

```python
pg = sns.PairGrid(titanic, hue= 'Survived')
pg.map(sns.scatterplot)
plt.show()
```
